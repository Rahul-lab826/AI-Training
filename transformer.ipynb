```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis")
classifier("This is my worst day in my life but a great day to live.")
```

```python
classifier=pipeline("zero-shot-classification")
classifier(
    "iI like to spend more money on travelling",
    candidate_labels=["technology", "politics", "business"],
)
```

```python
generator=pipeline("text-generation")
generator("Yesterday, I came through a forest")
```

```python
generator=pipeline("text-generation", model="Qwen/Qwen3-0.6B", trust_remote_code=True)
generator(
    "I am late to college so",
    max_length=30,
    num_return_sequences=2,
)
```

```python
generator=pipeline("fill-mask")
generator("The capital of France is <mask>.")
```
